## List Comprehension

```python
a = []
for i in range (31):
    if i % 2 == 0:
        a.append(i)

for i in range(len(a)): 
    print(a[i], end = " ")

#comprehension, puro bisoyt 1 line ei
b = [i for i in range(31) if i % 2 == 0]
print()
for i in range(len(a)): 
    print(b[i], end = " ")
```

## Sets in python

```python
s = {1,5,2,8,7}
print(type(s), s)

a = {}
print(type(a))

a = set()
print(type(a))

a.add(7)
a.add(1)
a.add(3)
a.add(2)
a.add(5)

print(a)

a.pop()
print(a)

for i in s:
    print(i)

a = {1,2,3}
b = {1,2,3,4,5,6}

c = a.union(b)
print(c)
d = a.intersection(b)
print(d)

print(a.isdisjoint(b))
print(a.issubset(b))
```

## Dictionary in python

```python
a = {}
print(type(a))

a = {
    "name" : "tanmoy",
    "age" : 25,
    "phone" : 1558
}
print(a)

a = {
    "gender" : "male"
}
print(a)

a = {
    "name" : "tanmoy",
    "age" : 25,
    "phone" : 1558,
    "age" : 30 # dui same namer key thakle update take store kore
}
print(a)

print(a["name"]) #eivabeo ekta single element access kora jay
print(a.get("age")) #eivabeo kora jay
```

## Dictonary Methods

```python
a = {
    "name" : "tanmoy",
    "age" : 25,
    "phone" : 1558
}
print(a)
print(a.get("math_marks"))
print(a.get("math_marks", 0)) 
# eikhane onno kichuo dite partam 
# 0 er jaigay jah kushi 
# mane eida nah paile oida kaj korbe erokom

a.update({"age" : 30})
print(a)

a["math_marks"] = 81
print(a)

del a["math_marks"]
print(a)

b = a.copy()
print(b)
```

## Ductionary Key, Value, Item 

```python
a = {
    "name" : "tanmoy",
    "age" : 25,
    "phone" : 1558
}

x = a.keys()
print(x)

a["gender"] = "male"

print(x)

y = a.values()
print(y)

z = a.items()
print(z)

for key,value in a.items():
    print(key, value)
```
