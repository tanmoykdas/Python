## Digit Sum


```python
a = int(input())
b = int(input())

a = a % 10
b = b % 10

print(a + b)

# taking input in in two line
```

```python
x = input()
n = x.split()

a = int(n[0])
b = int(n[1])

a = a % 10
b = b % 10

print(a + b)
# take two input in a same line
```

## Min Max

```python
n = list(map(int, input().split()))
for i in range(len(n)):
    print(n[i])
print("Min: ", min(n))
print("Max: ", max(n))

# 1 2 3 4 5
```

## Even Odd Positive Negative count

```python
a = list(map(int, input().split()))

e = 0; o = 0; p = 0; n = 0
for i in range(len(a)):
    if (a[i] >= 0): p += 1
    else : n += 1

    if (a[i] % 2): o += 1
    else : e += 1
print("even: ", e, " ", "odd: ", o, " ", "negative: ", n, " ", "positive: ", p)

# 1 2 -3 4 7
     
```

## Qdigit

```python
a = input()
a = int(a)
while(a):
    print(a % 10, end = " ")
    a //= 101234
# taken input 1234
```
